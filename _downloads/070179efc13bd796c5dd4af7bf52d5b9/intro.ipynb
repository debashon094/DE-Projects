```python
import torch
import numpy as np
```

```python
data = [[1,2],[3,4]]
x_data = torch.tensor(data)
print(x_data)
```

```python
one_like = torch.ones_like(x_data)
x_rand = torch.rand_like(x_data, dtype=torch.float)
print(one_like)
print(x_rand)
```

```python
shape = (2,3)
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)
print(rand_tensor)
print(ones_tensor)
print(zeros_tensor)
```
